# Random Forest HPO — Chạy trực tuyến trên Google Colab

**Không cần cài Python hay VS Code.**

1. Bấm **Runtime → Run all** (hoặc chạy từng ô bằng nút ▶).
2. Khi được yêu cầu, tải lên **file ZIP gốc `RandomForest_HPO_GitHub_Portfolio(1).zip`** của bạn.
3. Colab sẽ cài thư viện, chạy 30 trials, hiện bảng và biểu đồ, và tạo ZIP kết quả mới để tải về.

> Phải giữ tab Colab hoạt động cho đến khi chương trình chạy xong. Kết quả chính thức phải lấy từ lần chạy của bạn, không phải các số có sẵn trong ZIP.


```python
from google.colab import files
from pathlib import Path
import zipfile

print("Chọn file RandomForest_HPO_GitHub_Portfolio(1).zip trên máy của bạn")
uploaded = files.upload()
zip_names = [name for name in uploaded if name.lower().endswith('.zip')]
if len(zip_names) != 1:
    raise ValueError("Vui lòng chỉ upload đúng một file ZIP dự án.")
root = Path('/content/rf_hpo_run')
root.mkdir(exist_ok=True)
with zipfile.ZipFile(zip_names[0]) as z:
    for member in z.infolist():
        dest = (root / member.filename).resolve()
        if not dest.is_relative_to(root.resolve()):
            raise ValueError('ZIP chứa đường dẫn không an toàn')
    z.extractall(root)
projects = list(root.rglob('src/run_experiments.py'))
if len(projects) != 1:
    raise FileNotFoundError('Không tìm thấy src/run_experiments.py hoặc dự án bị trùng.')
project = projects[0].parent.parent
print('Đã tìm thấy thư mục dự án:', project)
```

```python
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(project / 'requirements.txt')], check=True)
print('Đã cài xong các thư viện.')
```

```python
import subprocess, sys
print('Bắt đầu chạy Random Forest HPO (30 trials)...')
subprocess.run([sys.executable, '-u', str(project/'src'/'run_experiments.py')], cwd=str(project), check=True)
print('ĐÃ CHẠY XONG!')
```

```python
import pandas as pd
from IPython.display import display, Image
display(pd.read_csv(project/'results'/'comparison.csv'))
print('Các file kết quả:')
for p in sorted((project/'results').iterdir()): print(' -',p.name)
for name in ['convergence.png','comparison.png','tradeoff.png']:
    print('\nBiểu đồ:',name)
    display(Image(filename=str(project/'figures'/name)))
```

```python
import shutil
from google.colab import files
output = shutil.make_archive('/content/RandomForest_HPO_DA_CHAY', 'zip', root_dir=project.parent, base_dir=project.name)
print('Tải ZIP dự án có kết quả đã chạy:', output)
files.download(output)
```
